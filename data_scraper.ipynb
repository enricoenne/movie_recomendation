```python
from movies import *
import pandas as pd
from tqdm import tqdm

import warnings

#warnings.simplefilter(action='ignore', category=pd.errors.PerformanceWarning)
```

### starting from 1 user

'enesidemo'

```python
visited_users = ['enesidemo']
visited_movies = []

big_df = pd.DataFrame().astype(pd.SparseDtype("int8", fill_value=0))

user = 'enesidemo'

user_search_sparse(user, big_df)
```

### loop through new movies

```python
movies_to_check = set(big_df.columns) - set(visited_movies)

for m in tqdm(movies_to_check):
    movie_search_sparse(m, big_df)
    visited_movies.append(m)
```

### save tables

```python
big_df.to_pickle("results.pkl")

pd.DataFrame(visited_users).to_csv('visited_users.csv', index=False)
pd.DataFrame(visited_movies).to_csv('visited_movies.csv', index=False)
```
