```python
from movies import *
from time import sleep

from tqdm import tqdm
```

```python
user_df = pd.read_pickle('users_df.pkl')
print(user_df.shape)

filtered_df = user_df.loc[:, user_df.astype(bool).sum(axis=0) >= 10]
print(filtered_df.shape)
```

```python
movies = filtered_df.columns
print(len(movies))
```

```python
metadata_df = pd.DataFrame()
visited_movies = []
```

```python
movies_to_check = set(movies) - set(visited_movies)

for m in tqdm(movies_to_check):
    current_metadata = get_movie_metadata(m)
    current_movie_df = pd.DataFrame([current_metadata])

    if metadata_df.empty:
        metadata_df = current_movie_df  # first row
    else:
        metadata_df = pd.concat([metadata_df, current_movie_df], ignore_index=True)
    
    visited_movies.append(m)

    sleep(0.2)
```

```python
metadata_df.to_csv('metadata.csv', index = False)
```
